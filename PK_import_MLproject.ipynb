**Step # 04: Let's handle null values and prepare the data for the ML model.**

- first [CLASS TYPE] feature
1. Categorical data and rename the column name, it has a space
2. Total missing values = 2,295
3. We cannot remove all null values because it may cause issues in the dataset
4. Replace all null values with the keyword "unknown"

```python
df_import1 = df_import.rename(columns={"CLASS TYPE": "CLASS_TYPE"})
```

```python
df_import1
```

```python
df_import1['CLASS_TYPE']=df_import1['CLASS_TYPE'].fillna('unknown')
```

```python
df_import1['CLASS_TYPE'].isnull().sum()
```

```python
df_import1.head(20)
```

- second, we have the [QUANTITY] and [%QUANTITY] features:
1. Both features contain missing values and are important for the dataset.
2. Total missing values: [QUANTITY] = 1,477 and [%QUANTITY] = 1,479
3. We need to replace the missing values using groupby on the [PRODUCT] feature.
4. The [PRODUCT] feature is required to estimate/derive quantity values.

```python
# QUANTITY
df_import1['QUANTITY'] = df_import1.groupby(['PRODUCT', 'YEAR'])['QUANTITY']\
                                 .transform(lambda x: x.fillna(x.median()))
```

```python
df_import1['QUANTITY'] = df_import1['QUANTITY'].fillna(df_import1['QUANTITY'].median())

# % QUANTITY
df_import1['% QUANTITY'] = df_import1['% QUANTITY'].fillna(0)
```

```python
df_import1.head(20)
```

- third [UNIT] feature
1. Categorical data
2. Total missimg Values = 1,479
3. It has three values (MT, KG, NO), here we have to handle data
4. Prepare the dataset for the ML Model
5. Best for ML (separating the data into [df_weight= for ML, df_count= separate analysis])

```python
df_import1['UNIT_TYPE'] = df_import1['UNIT'].map({
    'MT': 'WEIGHT',
    'KG': 'WEIGHT',
    'NO': 'COUNT'
})
```

### Imputing Missing `UNIT_TYPE` based on `PRODUCT`
("Imputed missing 'UNIT_TYPE' values based on product mode, and then global mode if necessary.")

```python
# Impute UNIT_TYPE based on the mode within each PRODUCT group
df_import1['UNIT_TYPE'] = df_import1.groupby('PRODUCT')['UNIT_TYPE'].transform(lambda x: x.fillna(x.mode()[0]) if not x.mode().empty else x)

# Handle any remaining NaN values (e.g., if an entire product group had no UNIT_TYPE defined)
# We'll fill these with the overall mode of UNIT_TYPE
if df_import1['UNIT_TYPE'].isnull().any():
    global_unit_type_mode = df_import1['UNIT_TYPE'].mode()[0]
    df_import1['UNIT_TYPE'].fillna(global_unit_type_mode, inplace=True)
```
